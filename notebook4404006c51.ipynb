```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Data
df = pd.read_csv('/kaggle/input/datasets/reemalmutairi1/supermarket-clean/supermarket_clean.csv')

# SQL Results as DataFrames
branch_profit = df.groupby('Branch')['Profit'].sum().round(2).sort_values(ascending=False)
category_profit = df.groupby('Category')['Profit'].sum().round(2).sort_values(ascending=False)
payment_profit = df.groupby('Payment_Type')['Profit'].sum().round(2).sort_values(ascending=False)

print("Data Loaded Successfully!")
print(df.shape)
```

```python
# Step 7: Dashboard - All Charts Together
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Chart 1: Branch
branch_profit.plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Total Profit by Branch')
axes[0].set_xlabel('Branch')
axes[0].set_ylabel('Total Profit ($)')
axes[0].tick_params(axis='x', rotation=45)
# Chart 2: Category
category_profit.plot(kind='bar', ax=axes[1], color='coral')
axes[1].set_title('Total Profit by Category')
axes[1].set_xlabel('Category')
axes[1].tick_params(axis='x', rotation=45)

# Chart 3: Payment Type
payment_profit.plot(kind='pie', ax=axes[2], autopct='%1.1f%%')
axes[2].set_title('Payment Type Distribution')
axes[2].set_ylabel('')
plt.tight_layout()
plt.show()
```
